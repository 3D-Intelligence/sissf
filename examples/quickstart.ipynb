```python
import json
from sissf import SceneState, Architecture, ModelInstance
from sissf.architecture.architecture import ArchRoom, ArchFloor
from sissf.geometry import Transform, Point3D

# Initialize scene
scene = SceneState(id="demo-scene")

# Create architecture
arch = Architecture(id="demo-arch")

# Create a room
room = ArchRoom(id="living_room")
arch.add_room(room)

# Add a floor (required for placing objects)
floor = ArchFloor(
    id="floor_living_room",
    points=[Point3D(0, 0, 0), Point3D(5, 0, 0), Point3D(5, 0, 4), Point3D(0, 0, 4)],
    depth=0.05,  # 5cm
    room_id="living_room",
)
arch.add_floor(floor)
scene.architecture = arch

# Add sofa to living room
sofa = ModelInstance(
    id="sofa-0",
    model_id="holodeck.sofa_example",  # Format: source.asset_id
    transform=Transform.from_rts(
        rotation=[0.0, 0.0, 0.0, 1.0],  # Quaternion [x, y, z, w]
        translation=[2.5, 0.3, 3.5],  # Position [x, y, z] in meters
        scale=[4, 0.6, 1.6],  # Scale [sx, sy, sz]
    ),
    parent_id="floor_living_room",
)
scene.add_instance(sofa)

# Save scene to JSON
with open("my_scene.json", "w") as f:
    json.dump(scene.to_dict(), f, indent=4)
```
